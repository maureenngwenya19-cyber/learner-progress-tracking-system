#  Learner Management

```python
class Learner:
    def __init__(self, learner_id, name, age, course,marks=None):
        self.learner_id = learner_id
        self.name = name
        self.age = age
        self.course = course
        self.marks=marks if marks is not None else []

    def __str__(self):
        return (
            f"Learner ID: {self.learner_id}\n"
            f"Name: {self.name}\n"
            f"Age: {self.age}\n"
            f"Course: {self.course}\n"
            f"Marks: {self.marks}"
        )
    # Example usage
learner1 = Learner(101, "John", 20, "Python",[12,13,15])
print(learner1)
```

```python
class LearnerManagementSystem:
    def __init__(self):
        self.learners = {}

          # Adds a new learner to the system. Prints an error if the learner ID already exists.

    def add_learner(self, learner_id, name, age, course, marks=None):
        if learner_id in self.learners:
            print(f"Error: Learner with ID {learner_id} already exists.")
            return
        new_learner = Learner(learner_id, name, age, course, marks)
        self.learners[learner_id] = new_learner
        print(f"Learner {name} added successfully.")

        # Displays the details of a specific learner. Prints an error if the learner is not found.

    def view_learner_details(self, learner_id):
        if learner_id not in self.learners:
            print(f"Error: Learner with ID {learner_id} not found.")
            return
        learner = self.learners[learner_id]
        print("\n--- Learner Details ---")
        print(f"Learner ID: {learner.learner_id}")
        print(f"Name: {learner.name}")
        print(f"Age: {learner.age}")
        print(f"Course: {learner.course}")
        print(f"Marks: {learner.marks if learner.marks else 'No marks recorded yet'}")
        print("-----------------------")

        #Updates the details of an existing learner. Prints an error if the learner is not found.

    def update_learner_details(self, learner_id, name=None, age=None, course=None):
        if learner_id not in self.learners:
            print(f"Error: Learner with ID {learner_id} not found.")
            return
        learner = self.learners[learner_id]
        if name:
            learner.name = name
        if age:
            learner.age = age
        if course:
            learner.course = course
        print(f"Learner {learner_id} details updated successfully.")

        #Removes a learner from the system. Prints an error if the learner is not found.

    def remove_learner(self, learner_id):
        if learner_id not in self.learners:
            print(f"Error: Learner with ID {learner_id} not found.")
            return
        del self.learners[learner_id]
        print(f"Learner with ID {learner_id} removed successfully.")

         #Adds a mark to a learner's record. Prints an error if the learner is not found.

    def add_marks_to_learner(self, learner_id, mark):
       
        if learner_id not in self.learners:
            print(f"Error: Learner with ID {learner_id} not found.")
            return
        self.learners[learner_id].marks.append(mark)
        print(f"Mark {mark} added to learner {learner_id}.")

# Example Usage
if __name__ == "__main__":
    system = LearnerManagementSystem()

    # Adding learners
    system.add_learner(101, "Alice Smith", 20, "Computer Science", [85, 90])
    system.add_learner(102, "Bob Johnson", 22, "Physics", [78, 88, 95])
    system.add_learner(101, "Duplicate Alice", 21, "Math") # Example of error handling for duplicate ID

    # Viewing learner details
    system.view_learner_details(101)
    system.view_learner_details(103) 

    # Updating learner details
    system.update_learner_details(102, age=23, course="Chemistry")
    system.view_learner_details(102)

    # Adding marks to a learner
    system.add_marks_to_learner(101, 92)
    system.view_learner_details(101)

    # Removing a learner
    system.remove_learner(102)
    system.view_learner_details(102) 
```

# Inheritance

```python

# Base class for Person
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Derived class for Learner
class Learner(Person):
    def __init__(self, learner_id, name, age, course):
        super().__init__(name, age)
        self.learner_id = learner_id
        self.course = course
        self._marks = []

  # Adds a mark to the learner's record
    def add_mark(self, mark):
        if isinstance(mark, (int, float)) and 0 <= mark <= 100:
            self._marks.append(mark)
        else:
            print("Invalid mark. Marks must be between 0 and 100.")

               #Calculates and returns the average mark.
    def get_average_mark(self):
        if not self._marks:
            return 0
        return sum(self._marks) / len(self._marks)

# Example usage:
learner1 = Learner("L001", "Alice", 20, "Python Programming")
learner1.add_mark(85)
learner1.add_mark(92)
learner1.add_mark(78)

print(f"Learner ID: {learner1.learner_id}")
print(f"Name: {learner1.name}")
print(f"Age: {learner1.age}")
print(f"Course: {learner1.course}")
print(f"Marks: {learner1._marks}")
print(f"Average Mark: {learner1.get_average_mark():.2f}")

# Example of invalid mark
learner1.add_mark(105)
```

#  Encapsulation

```python
def add_mark(self, mark):
        if 0 <= mark <= 100:
            self._marks.append(mark)
        else:
            raise ValueError("Invalid mark entry. Marks must be between 0 and 100.")

def get_average(self):
        if not self._marks:
            return 0
        return self._sum_marks(self._marks) / len(self._marks)

def _sum_marks(self, marks):
        if not marks:
            return 0
        return marks[0] + self._sum_marks(marks[1:])

# Example usage
learner1 = Learner("L001", "Sarah",17,'Python')

learner1.add_mark(75)
learner1.add_mark(85)
learner1.add_mark(90)

print("Learner 1 average:", learner1.get_average_mark())

learner2= Learner("L002", "Jones",18,'Python')

learner2.add_mark(95)
learner2.add_mark(65)
learner2.add_mark(50)

print("Learner 2 average:", learner2.get_average_mark())
```

```python

import tkinter as tk
from tkinter import messagebox

# Hide main tkinter window
root = tk.Tk()
root.withdraw()

class Learner:
    def __init__(self, learner_id, name, age, course="General Studies"):
        self.learner_id = learner_id
        self.name = name
        self.age = age
        self.course = course
        self.marks = []
        self.max_marks_allowed = 3 # Limit to 3 marks

    def add_mark(self, mark):
        if 0 <= mark <= 100 and len(self.marks) < self.max_marks_allowed:
            self.marks.append(mark)
            return True
        return False

    def get_average(self):
        if not self.marks:
            return 0
        # Requirement 8: Recursive function to sum marks
        total = self.recursive_sum(self.marks)
        return total / len(self.marks)

    def recursive_sum(self, marks_list):
        if not marks_list:
            return 0
        return marks_list[0] + self.recursive_sum(marks_list[1:])

learners = []

# Requirement 7: Predicate function
def qualifies_for_certificate(learner):
    return learner.get_average() >= 50

# Requirement 7: Function with default argument
def add_learner(learner_id, name, age, course="General Studies"):
    new_learner = Learner(learner_id, name, age, course)
    learners.append(new_learner)
    messagebox.showinfo("Success", f"Learner {name} registered successfully!")

def find_learner(learner_id):
    for learner in learners:
        if learner.learner_id == learner_id:
            return learner
    return None

def capture_marks(learner):
    while len(learner.marks) < learner.max_marks_allowed:
        try:
            entry = input(f"Enter mark ({len(learner.marks)}/{learner.max_marks_allowed}) (or 'done' to finish): ")
            if entry.lower() == 'done':
                break
            mark = float(entry)
            if learner.add_mark(mark):
                print(f"Mark {mark} added.")
            else:
                if len(learner.marks) >= learner.max_marks_allowed:
                    messagebox.showwarning("Limit Reached", f"You have already added {learner.max_marks_allowed} marks for this learner.")
                    break
                else:
                    messagebox.showwarning("Invalid Input", "Mark must be between 0 and 100.")
        except ValueError:
            messagebox.showerror("Error", "Please enter a valid numeric mark.")
    if len(learner.marks) < learner.max_marks_allowed:
        print(f"Mark capture finished. {learner.max_marks_allowed - len(learner.marks)} marks remaining.")


def display_summary(learner):
    avg = learner.get_average()
    qualifies = qualifies_for_certificate(learner)

    if avg < 50:
        result = "Fail"
    elif avg < 75:
        result = "Pass"
    else:
        result = "Pass with Distinction"

    cert_status = "Yes" if qualifies else "No"

    print(f"\n--- Learner Summary ---")
    print(f"ID: {learner.learner_id} | Name: {learner.name}")
    print(f"Course: {learner.course}")
    print(f"Marks: {learner.marks}")
    print(f"Average: {avg:.2f}")
    print(f"Result: {result}")
    print(f"Qualifies for Certificate: {cert_status}")
    messagebox.showinfo("Result", f"{learner.name}: {result}\nCert: {cert_status}")

# Requirement 6: Main Menu Loop
while True:
    print("\n1. Add Learner\n2. View Summary\n3. Capture Marks\n4. Exit")
    choice = input("Choose an option: ")

    try:
        if choice == '1':
            l_id = input("Enter Learner ID: ")
            name = input("Enter Learner Name: ")
            # Requirement 9: Exception handling for numeric input
            age = int(input("Enter Learner Age: "))
            if age < 0: raise ValueError("Age cannot be negative.")
            course = input("Enter Learner Course: ")

            # Requirement 7: Keyword argument call
            add_learner(learner_id=l_id, name=name, age=age, course=course)
    
        elif choice == '2':
            l_id = input("Enter Learner ID to capture marks: ")
            learner = find_learner(l_id)
            if learner:
                capture_marks(learner)
            else:
                print("Learner not found.")
                
        elif choice == '3':
            l_id = input("Enter Learner ID to view: ")
            learner = find_learner(l_id)
            if learner:
                display_summary(learner)
            else:
                print("Learner not found.")        

        elif choice == '4':
            if messagebox.askyesno("Exit", "Are you sure you want to quit?"):
                break
        else:
            print("Invalid selection.")

    except ValueError as e:
        messagebox.showerror("Input Error", f"Invalid data entered: {e}")
    except Exception as e: # Catch any other unexpected errors
        messagebox.showerror("An Error Occurred", f"An unexpected error occurred: {e}")

```
